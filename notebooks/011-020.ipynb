### 11. Create a 3x3 identity matrix (★☆☆)

````python
Z = np.eye(3)
print(Z)

# same result, other ways:
print(np.identity(3))
print(np.diag([1, 1, 1]))
````

**Concept:** `np.eye(n)` (or `np.identity(n)`) builds an identity matrix: 1s on the main diagonal, 0s elsewhere. Default dtype is float64.

### 12. Create a 3x3x3 array with random values (★☆☆)

````python
rng = np.random.default_rng(42)     # modern generator; seed makes results reproducible
Z = rng.random((3, 3, 3))
print(Z)
print(Z.shape)                      # (3, 3, 3)

# legacy style you'll see in older tutorials:
# Z = np.random.random((3, 3, 3))
````

**Concept:** `rng.random(shape)` gives uniform values in [0, 1). Pass the shape as a tuple. Use `default_rng(seed)` for reproducible results (the modern API; `np.random.random` is the older style).

### 13. Create a 10x10 array with random values and find the minimum and maximum values (★☆☆)

````python
rng = np.random.default_rng(42)
Z = rng.random((10, 10))

Zmin, Zmax = Z.min(), Z.max()
print(Zmin, Zmax)

# where are they?
print(np.unravel_index(Z.argmin(), Z.shape))
print(np.unravel_index(Z.argmax(), Z.shape))
````

**Concept:** `.min()` / `.max()` reduce the whole array to one value. `.argmin()` / `.argmax()` return the position in the *flattened* array; `np.unravel_index` converts it back to (row, col).

### 14. Create a random vector of size 30 and find the mean value (★☆☆)

````python
rng = np.random.default_rng(42)
Z = rng.random(30)

m = Z.mean()
print(m)

# same thing:
print(np.mean(Z))
````

**Concept:** `.mean()` averages all elements. With multi-dimensional arrays use the `axis` argument (`axis=0` averages down columns, `axis=1` across rows).

### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆)

````python
Z = np.ones((10, 10))
Z[1:-1, 1:-1] = 0       # everything except the first/last row and column
print(Z)
````

**Concept:** slice assignment. `1:-1` means "from the second element up to, but not including, the last", so `Z[1:-1, 1:-1]` selects the interior. Fill everything with 1, then zero out the inside.

### 16. How to add a border (filled with 0's) around an existing array? (★☆☆)

````python
Z = np.ones((5, 5))
Z = np.pad(Z, pad_width=1, mode="constant", constant_values=0)
print(Z)
print(Z.shape)          # (7, 7)
````

**Concept:** `np.pad(array, pad_width, mode="constant", constant_values=0)` surrounds an array with a border. `pad_width=1` adds one layer on every side, so (5, 5) becomes (7, 7).

### 17. What is the result of the following expression? (★☆☆)
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
print(0 * np.nan)                # nan   (any arithmetic with nan stays nan)
print(np.nan == np.nan)          # False (nan is not equal to anything, even itself)
print(np.inf > np.nan)           # False (every comparison with nan is False)
print(np.nan - np.nan)           # nan
print(np.nan in set([np.nan]))   # True  (Python checks identity `is` before `==`)
print(0.3 == 3 * 0.1)            # False (floating point: 3 * 0.1 is 0.30000000000000004)

# the right way to check:
print(np.isnan(np.nan))                  # True
print(np.isclose(0.3, 3 * 0.1))          # True
print(3 * 0.1)
````

**Concept:** `nan` propagates through arithmetic and fails every comparison, so test it with `np.isnan`. `np.nan in set([np.nan])` is True only because it is the *same object* (identity check). Floats are inexact, so compare with `np.isclose`, never `==`.

### 18. Create a 5x5 matrix with values 1,2,3,4 just below the diagonal (★☆☆)

````python
Z = np.diag(1 + np.arange(4), k=-1)
print(Z)

# k=-1 -> one diagonal below the main one; k=+1 would be above
````

**Concept:** `np.diag(v, k)` builds a matrix with vector `v` on the k-th diagonal (`k=0` main, negative below, positive above). The matrix size is `len(v) + abs(k)`, so 4 values with `k=-1` gives 5x5.

### 19. Create a 8x8 matrix and fill it with a checkerboard pattern (★☆☆)

````python
Z = np.zeros((8, 8), dtype=int)
Z[1::2, ::2] = 1        # odd rows, even columns
Z[::2, 1::2] = 1        # even rows, odd columns
print(Z)

# alternative: a cell is 1 when row + col is odd
i, j = np.indices((8, 8))
print((i + j) % 2)
````

**Concept:** slicing with a step. `1::2` selects every second element starting at index 1; `::2` starts at 0. Two slice assignments produce the pattern. Alternative: `np.indices` + `(row + col) % 2`.

### 20. Consider a (6,7,8) shape array, what is the index (x,y,z) of the 100th element? (★☆☆)

````python
print(np.unravel_index(99, (6, 7, 8)))   # (1, 5, 3)

# the 100th element is index 99 (counting starts at 0)
# check by hand: 99 = 1*(7*8) + 5*8 + 3  ->  56 + 40 + 3 = 99
````

**Concept:** `np.unravel_index(flat_index, shape)` converts a position in the flattened array to a multi-dimensional index. The 100th element has flat index **99** because counting starts at 0. The inverse is `np.ravel_multi_index`.
