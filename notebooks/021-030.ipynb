### 21. Create a checkerboard 8x8 matrix using the tile function (★☆☆)

````python
Z = np.tile(np.array([[0, 1], [1, 0]]), (4, 4))
print(Z)
print(Z.shape)          # (8, 8)
````

**Concept:** `np.tile(block, reps)` repeats a small array. A 2x2 block tiled `(4, 4)` times gives 8x8 (2×4 per axis), which makes the checkerboard from problem 19 without slicing.

### 22. Normalize a 5x5 random matrix (★☆☆)

````python
rng = np.random.default_rng(42)
Z = rng.random((5, 5))

# z-score normalization: mean 0, standard deviation 1
Zn = (Z - Z.mean()) / Z.std()
print(Zn)
print(np.isclose(Zn.mean(), 0), Zn.std())   # True 1.0

# min-max normalization: values scaled to the range [0, 1]
Zm = (Z - Z.min()) / (Z.max() - Z.min())
print(Zm.min(), Zm.max())
````

**Concept:** "normalize" can mean two things. Z-score: `(Z - mean) / std` gives mean 0 and std 1. Min-max: `(Z - min) / (max - min)` rescales to [0, 1]. Both use broadcasting, so a scalar is applied to every element with no loop.

### 23. Create a custom dtype that describes a color as four unsigned bytes (RGBA) (★☆☆)

````python
color = np.dtype([("r", np.ubyte),
                  ("g", np.ubyte),
                  ("b", np.ubyte),
                  ("a", np.ubyte)])

print(color)
print(color.itemsize)               # 4 bytes: one per channel

# using it:
pixels = np.array([(255, 0, 0, 255), (0, 128, 255, 128)], dtype=color)
print(pixels)
print(pixels["r"])                  # access one field across all pixels
````

**Concept:** a *structured dtype* is a list of `(name, type)` pairs, so one array element holds several named fields. `np.ubyte` is an unsigned 8-bit integer (0-255), and four of them make 4 bytes. Access a field with `array["name"]`.

### 24. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product) (★☆☆)

````python
A = np.arange(15).reshape(5, 3)
B = np.arange(6).reshape(3, 2)

Z = np.dot(A, B)            # matrix product
print(Z)
print(Z.shape)              # (5, 2)

# same thing with the @ operator:
print(A @ B)

# NOT the same: * is element-wise and fails here because shapes differ
# A * B  -> ValueError
````

**Concept:** matrix product needs inner dimensions to match: (5, 3) @ (3, 2) gives (5, 2). Use `np.dot` or the `@` operator. `*` is element-wise multiplication, not the matrix product.

### 25. Given a 1D array, negate all elements which are between 3 and 8, in place. (★☆☆)

````python
Z = np.arange(11)
print(Z)

Z[(3 < Z) & (Z < 8)] *= -1  # in place: modifies Z itself
print(Z)
````

**Concept:** combine conditions with `&` (and) / `|` (or), each wrapped in parentheses, to build a boolean mask. `Z[mask] *= -1` modifies the array in place. Python's `and` / `or` do not work on arrays. "Between 3 and 8" here means strictly between, so 4 to 7 are negated.

### 26. What is the output of the following script? (★☆☆)
```python
# Author: Jake VanderPlas

print(sum(range(5),-1))
from numpy import *
print(sum(range(5),-1))
```

````python
import builtins

print(builtins.sum(range(5), -1))   # 9  : Python's sum -> 0+1+2+3+4 plus start value -1
print(np.sum(range(5), -1))         # 10 : NumPy's sum -> -1 is the *axis*, not a start value

# Running the original script in an isolated namespace (so the star-import
# does not shadow the built-in sum/min/max/abs in this notebook):
script = '''
print(sum(range(5), -1))
from numpy import *
print(sum(range(5), -1))
'''
exec(script, {})                    # prints 9 then 10
````

**Concept:** the built-in `sum(iterable, start)` treats `-1` as a starting value (so 10 - 1 = 9). After `from numpy import *`, `sum` becomes `np.sum(a, axis)`, where `-1` means the last axis (so the result is 10). Star imports shadow built-ins like `sum`, `min`, `max`, `abs`, `round`, so avoid `from numpy import *`.

### 27. Consider an integer vector Z, which of these expressions are legal? (★☆☆)
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```

````python
Z = np.arange(1, 6)

exprs = {
    "Z**Z":     lambda: Z**Z,
    "2 << Z >> 2": lambda: 2 << Z >> 2,
    "Z <- Z":   lambda: Z <- Z,
    "1j*Z":     lambda: 1j*Z,
    "Z/1/1":    lambda: Z/1/1,
    "Z<Z>Z":    lambda: Z<Z>Z,
}

for text, f in exprs.items():
    try:
        print(f"{text:12} -> {f()}")
    except Exception as e:
        print(f"{text:12} -> {type(e).__name__}: {e}")
````

**Concept:** all are legal except `Z<Z>Z`. `Z <- Z` is parsed as `Z < (-Z)`, not an arrow. `2 << Z >> 2` is `(2 << Z) >> 2` (bit shifts). `1j*Z` gives complex numbers and `Z/1/1` gives floats. `Z<Z>Z` is a chained comparison, which Python expands to `(Z<Z) and (Z>Z)`; `and` needs a single True/False, so NumPy raises a ValueError about ambiguous truth values.

### 28. What are the result of the following expressions? (★☆☆)
```python
np.array(0) / np.array(0)
np.array(0) // np.array(0)
np.array([np.nan]).astype(int).astype(float)
```

````python
with np.errstate(all="ignore"):         # hide the RuntimeWarnings so output stays clean
    print(np.array(0) / np.array(0))                       # nan
    print(np.array(0) // np.array(0))                      # 0
    print(np.array([np.nan]).astype(int).astype(float))    # platform dependent
````

**Concept:** `0 / 0` gives `nan` (with an "invalid value" warning). Integer floor division `0 // 0` gives `0` (with a "divide by zero" warning), because integers have no nan. Casting `nan` to int is undefined: the result depends on the platform (commonly the smallest int64, about -9.22e+18), so never rely on it.

### 29. How to round away from zero a float array ? (★☆☆)

````python
rng = np.random.default_rng(42)
Z = rng.uniform(-10, 10, 10)
print(Z)

# Method 1: ceil for positives, floor for negatives
r1 = np.where(Z > 0, np.ceil(Z), np.floor(Z))

# Method 2: ceil the absolute value, then copy the original sign back
r2 = np.copysign(np.ceil(np.abs(Z)), Z)

print(r1)
print(np.array_equal(r1, r2))   # True

# For comparison, np.round goes to the NEAREST integer, not away from zero:
print(np.round(Z))
````

**Concept:** "away from zero" means up for positives and down for negatives. Use `np.where(Z > 0, np.ceil(Z), np.floor(Z))`, or `np.copysign(np.ceil(np.abs(Z)), Z)`. Plain `np.round` rounds to the nearest integer, which is different. `np.where(cond, a, b)` is a vectorized if/else.

### 30. How to find common values between two arrays? (★☆☆)

````python
rng = np.random.default_rng(42)
Z1 = rng.integers(0, 10, 10)
Z2 = rng.integers(0, 10, 10)
print(Z1)
print(Z2)

print(np.intersect1d(Z1, Z2))       # sorted, unique common values

# Alternative: keep the elements of Z1 that appear in Z2 (duplicates kept)
print(Z1[np.isin(Z1, Z2)])
````

**Concept:** `np.intersect1d(a, b)` returns the sorted unique values found in both arrays. `np.isin(a, b)` returns a boolean mask of which elements of `a` appear in `b`, handy when you want to keep duplicates or positions.
