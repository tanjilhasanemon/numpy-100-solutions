### 3. Create a null vector of size 10 (★☆☆)

```python
Z = np.zeros(10)
print(Z)
print(Z.dtype)          # float64 by default

# integer version:
print(np.zeros(10, dtype=int))
```

### 4. How to find the memory size of any array (★☆☆)

```python
Z = np.zeros((10, 10))

print(Z.size)                         # number of elements -> 100
print(Z.itemsize)                     # bytes per element  -> 8 (float64)
print(f"{Z.size * Z.itemsize} bytes") # total memory       -> 800

# shortcut that gives the same number:
print(Z.nbytes)
```

**Concept:** memory = `size × itemsize`; `.nbytes` is the built-in shortcut. Changing dtype (e.g. `float32`) halves the memory.

### 6. Create a null vector of size 10 but the fifth value which is 1 (★☆☆)

```python
Z = np.zeros(10)
Z[4] = 1                # indexing starts at 0, so the 5th element is index 4
print(Z)
```

**Concept:** indexing is 0-based, so the 5th element is `Z[4]`. Arrays are mutable; assign directly with `Z[i] = value`.

### 7. Create a vector with values ranging from 10 to 49 (★☆☆)

```python
Z = np.arange(10, 50)   # stop value is exclusive, so use 50 to include 49
print(Z)
print(Z.shape)          # (40,)
```

**Concept:** `np.arange(start, stop)` excludes `stop` (like Python's `range`). To include 49, stop at 50.

### 8. Reverse a vector (first element becomes last) (★☆☆)

```python
Z = np.arange(50)
Z = Z[::-1]             # slice with step -1
print(Z)

# The reversed array is a VIEW, not a copy:
A = np.arange(5)
B = A[::-1]
B[0] = 99
print(A)                # A changed too -> [ 0  1  2  3 99]
```

**Concept:** slicing `[::-1]` reverses an array. It returns a *view* (shares memory with the original), so editing it edits the original. Use `.copy()` if you need independence.

### 9. Create a 3x3 matrix with values ranging from 0 to 8 (★☆☆)

```python
Z = np.arange(9).reshape(3, 3)
print(Z)
print(Z.shape)          # (3, 3)
```

**Concept:** create a flat range with `np.arange(9)`, then reshape. The new shape must hold the same number of elements (3×3 = 9). `reshape` fills row by row.

### 10. Find indices of non-zero elements from [1,2,0,0,4,0] (★☆☆)

```python
arr = [1, 2, 0, 0, 4, 0]
nz = np.nonzero(arr)
print(nz)               # (array([0, 1, 4]),)
print(nz[0])            # [0 1 4]

# Equivalent with a boolean mask:
a = np.array(arr)
print(np.where(a != 0)[0])
```

**Concept:** `np.nonzero(a)` returns a tuple of index arrays, one per dimension. For a 1D array take `[0]`. `np.where(condition)` with a single argument does the same thing.
